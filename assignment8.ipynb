### 1. Print First Character N Times

```python
word = input()
length = len(word)
first_char = word[0]

for i in range(length):
    print(first_char)
```

### 5. Sum of Natural Numbers

```python
n = int(input())
total_sum = 0
for i in range(1, n + 1):
    total_sum = total_sum + i
print(total_sum)
```

### 6. Average of Numbers from 1 to N

```python
n = int(input())
total_sum = 0
for i in range(1, n + 1):
    total_sum = total_sum + i

average = total_sum / n
print(average)
```

### 13. Styled Word (Hyphen Separated)

```python
word = input()
length = len(word)
styled_word = ""

for i in range(length):
    if i == (length - 1):
        styled_word = styled_word + word[i]
    else:
        styled_word = styled_word + word[i] + "-"

print(styled_word)
```
