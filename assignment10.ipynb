### 1. Letter, Digit or Special Character

```javascript
// Note: Run in Node.js environment
const fs = require('fs');
const input = fs.readFileSync(0, 'utf-8').trim();

if (/[0-9]/.test(input)) {
    console.log("Digit");
} else if (/[a-z]/.test(input)) {
    console.log("Lowercase Letter");
} else if (/[A-Z]/.test(input)) {
    console.log("Uppercase Letter");
} else {
    console.log("Special Character");
}
```

### 2. String Index Characters

```javascript
const fs = require('fs');
const input = fs.readFileSync(0, 'utf-8').trim().split('\n');

const s = input[0];
for (let i = 1; i <= s.length; i++) {
    const idx = parseInt(input[i]);
    if (!isNaN(idx)) {
        console.log(s[idx]);
    }
}
```

### 3. Prefix String Pattern

```javascript
const fs = require('fs');
const s = fs.readFileSync(0, 'utf-8').trim();

for (let i = 1; i <= s.length; i++) {
    console.log(s.slice(0, i));
}
```

### 4. Number Grid Pattern

```javascript
const fs = require('fs');
const n = parseInt(fs.readFileSync(0, 'utf-8').trim());

for (let i = 0; i < n; i++) {
    console.log((i + " ").repeat(n).trim());
}
```

### 5. Shuffle String by Index

```javascript
const fs = require('fs');
const input = fs.readFileSync(0, 'utf-8').trim().split('\n');

const s = input[0];
let shuffled = "";

for (let i = 1; i <= s.length; i++) {
    const idx = parseInt(input[i]);
    if (!isNaN(idx)) {
        shuffled += s[idx];
    }
}

console.log(shuffled);
```

### 6. Repeated Rows Pattern

```javascript
const fs = require('fs');
const input = fs.readFileSync(0, 'utf-8').trim().split('\n');

const a = parseInt(input[0]);
const b = parseInt(input[1]);

let row = [];
for (let i = a; i < a + b; i++) {
    row.push(i);
}

const rowStr = row.join(" ");

for (let i = 0; i < b; i++) {
    console.log(rowStr);
}
```

### 7. Count of Zeroes

```javascript
const fs = require('fs');
const s = fs.readFileSync(0, 'utf-8').trim();

const count = (s.match(/0/g) || []).length;

if (count > 3) {
    console.log("Count of zeroes is greater than three");
} else {
    console.log("Count of zeroes is not greater than three");
}
```

### 8. Number of Even Digits

```javascript
const fs = require('fs');
const s = fs.readFileSync(0, 'utf-8').trim();

let count = 0;
for (let char of s) {
    if (/[0-9]/.test(char)) {
        if (parseInt(char) % 2 === 0) {
            count++;
        }
    }
}

if (count > 2) {
    console.log("Count of even digits is greater than two");
} else {
    console.log("Count of even digits is not greater than two");
}
```

### 9. Prime Number

```javascript
const fs = require('fs');
const n = parseInt(fs.readFileSync(0, 'utf-8').trim());

let isPrime = true;
if (n < 2) {
    isPrime = false;
} else {
    for (let i = 2; i <= Math.sqrt(n); i++) {
        if (n % i === 0) {
            isPrime = false;
            break;
        }
    }
}

if (isPrime) {
    console.log("Prime Number");
} else {
    console.log("Not a Prime Number");
}
```

### 10. Star Increasing Pattern

```javascript
const fs = require('fs');
const n = parseInt(fs.readFileSync(0, 'utf-8').trim());

for (let i = 0; i < n; i++) {
    console.log("* ".repeat(1 + 2 * i).trim());
}
```

### 11. Reverse Star Pattern

```javascript
const fs = require('fs');
const n = parseInt(fs.readFileSync(0, 'utf-8').trim());

for (let i = n; i > 0; i--) {
    console.log("* ".repeat(i).trim());
}
```

### 12. Hollow Square

```javascript
const fs = require('fs');
const n = parseInt(fs.readFileSync(0, 'utf-8').trim());

const topBottom = "+ " + "- ".repeat(n - 2) + "+";
const middle = "| " + "  ".repeat(n - 2) + "|";

console.log(topBottom);
for (let i = 0; i < n - 2; i++) {
    console.log(middle);
}
console.log(topBottom);
```

### 13. Case Conversion

```javascript
const fs = require('fs');
const s = fs.readFileSync(0, 'utf-8').trim();

let result = "";
for (let i = 0; i < s.length; i++) {
    const char = s[i];
    if (char === char.toUpperCase() && /[A-Z]/.test(char)) {
        if (i !== 0) {
            result += "-";
        }
        result += char.toLowerCase();
    } else {
        result += char;
    }
}

console.log(result);
```

### 14. Square Numbers Pattern

```javascript
const fs = require('fs');
const n = parseInt(fs.readFileSync(0, 'utf-8').trim());

for (let i = 1; i <= n; i++) {
    console.log((i + " ").repeat(n).trim());
}
```

### 15. Rectangular Pattern

```javascript
const fs = require('fs');
const input = fs.readFileSync(0, 'utf-8').trim().split('\n');

const m = parseInt(input[0]);
const n = parseInt(input[1]);

let row = [];
for (let i = 1; i <= n; i++) {
    row.push(i);
}

const rowStr = row.join(" ");

for (let i = 0; i < m; i++) {
    console.log(rowStr);
}
```

### 16. Pyramid Pattern

```javascript
const fs = require('fs');
const n = parseInt(fs.readFileSync(0, 'utf-8').trim());

for (let i = 0; i < n; i++) {
    const spaces = " ".repeat(n - 1 - i);
    const stars = "* ".repeat(1 + 2 * i).trim();
    console.log(spaces + stars);
}
```
