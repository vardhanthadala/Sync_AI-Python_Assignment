# Foundations of Programming: Loop-Based Problem Set

### 1. Next 10 Numbers

```python
n = int(input())
for i in range(1, 11):
    print(n + i)
```

### 2. Print Numbers from M to N

```python
m = int(input())
n = int(input())
for i in range(m, n + 1):
    print(i)
```

### 3. Sum of Numbers from M to N

```python
m = int(input())
n = int(input())
total_sum = 0
for i in range(m, n + 1):
    total_sum = total_sum + i
print(total_sum)
```

### 4. Square Star Pattern

```python
n = int(input())
for i in range(n):
    print("* " * n)
```

### 5. Number Square Pattern

```python
n = int(input())
for i in range(1, n + 1):
    print(str(i) * n)
```

### 6. Rectangle Star Pattern

```python
m = int(input())
n = int(input())
for i in range(m):
    print("*" * n)
```

### 7. Increasing Star Pattern

```python
n = int(input())
for i in range(1, n + 1):
    print("*" * i)
```

### 8. Number Triangle Pattern (with spaces)

```python
n = int(input())
for i in range(1, n + 1):
    row = (str(i) + " ") * i
    print(row.strip())
```

### 9. Mixed Pattern (* and +)

```python
n = int(input())
for i in range(1, n):
    row = "* " * i
    print(row.strip())

last_row = "+ " * n
print(last_row.strip())
```

### 10. Repeat Pattern Twice

```python
n = int(input())
for j in range(2):
    for i in range(1, n + 1):
        print(str(i) * i)
```

### 11. Number Rectangle

```python
m = int(input())
n = int(input())
for i in range(1, m + 1):
    row = (str(i) + " ") * n
    print(row.strip())
```

### 12. Print Numbers from 0 to N

```python
n = int(input())
for i in range(n + 1):
    print(i)
```

### 13. Average of N Numbers

```python
n = int(input())
total_sum = 0
for i in range(n):
    number = float(input())
    total_sum = total_sum + number

average = total_sum / n
print(average)
```

### 14. Diamond-like Star Pattern

```python
n = int(input())
for j in range(2):
    for i in range(1, n + 1):
        row = "* " * i
        print(row.strip())
```

### 15. Product of Numbers from M to N

```python
m = int(input())
n = int(input())
product = 1
for i in range(m, n + 1):
    product = product * i
print(product)
```

### 16. Sum of N Squares

```python
n = int(input())
total_sum = 0
for i in range(1, n + 1):
    total_sum = total_sum + (i * i)
print(total_sum)
```

### 17. Print Integers (Reverse)

```python
n = int(input())
for i in range(n, 0, -1):
    print(i)
```

### 18. Solid Rectangle using +

```python
m = int(input())
n = int(input())
for i in range(m):
    row = "+ " * n
    print(row.strip())
```

### 19. Double Triangle Pattern

```python
n = int(input())
for j in range(2):
    for i in range(1, n + 1):
        row = "* " * i
        print(row.strip())
```
