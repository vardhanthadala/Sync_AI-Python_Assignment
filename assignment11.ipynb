```python
# 1. Kth Largest Factor
n = int(input())
k = int(input())

factors = []
for i in range(1, n + 1):
    if n % i == 0:
        factors.append(i)

factors.sort(reverse=True)

if len(factors) >= k:
    print(factors[k - 1])
```

```python
# 2. First Prime Number
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

n = int(input())
found = False
for _ in range(n):
    val = int(input())
    if not found and is_prime(val):
        print(val)
        found = True
```

```python
# 3. Print Until Multiple of 5
n = int(input())
for _ in range(n):
    val = int(input())
    if val % 5 == 0:
        break
    print(val)
```

```python
# 4. Multiples of 3
n = int(input())
for _ in range(n):
    val = int(input())
    if val % 3 == 0:
        print(val)
```

```python
# 5. Alphabet Triangle
n = int(input())
for i in range(1, n + 1):
    spaces = " " * (2 * (n - i))
    chars = []
    for j in range(i):
        chars.append(chr(65 + j))
    print(spaces + " ".join(chars))
```

```python
# 6. Even and Odd Count
m = int(input())
n = int(input())
even_count = 0
odd_count = 0
for i in range(m, n + 1):
    if i % 2 == 0:
        even_count += 1
    else:
        odd_count += 1
print(even_count)
print(odd_count)
```

```python
# 7. Vowels and Consonants
s = input()
vowels = "aeiouAEIOU"
v_count = 0
c_count = 0
for char in s:
    if char.isalpha():
        if char in vowels:
            v_count += 1
        else:
            c_count += 1
print(v_count)
print(c_count)
```

```python
# 8. One Color
s = input()
r_count = s.count('R')
g_count = s.count('G')
print(min(r_count, g_count))
```

```python
# 9. Alphabetic Triangle
n = int(input())
for i in range(1, n + 1):
    chars = []
    for j in range(i):
        chars.append(chr(65 + j))
    print(" ".join(chars))
```

```python
# 10. Right Triangle Numbers
n = int(input())
for i in range(1, n + 1):
    row = ""
    for j in range(1, i + 1):
        row += str(j)
    for j in range(i - 1, 0, -1):
        row += str(j)
    print(row)
```

```python
# 11. Case Conversion
s = input()
result = ""
for i in range(len(s)):
    if s[i].isupper():
        if i != 0:
            result += "_"
        result += s[i].lower()
    else:
        result += s[i]
print(result)
```

```python
# 12. Sum of Non-Primes
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

n = int(input())
total_sum = 0
for _ in range(n):
    val = int(input())
    if not is_prime(val):
        total_sum += val
print(total_sum)
```

```python
# 13. Sandglass Pattern
n = int(input())
# Top half
for i in range(n, 0, -1):
    spaces = " " * (n - i)
    stars = " ".join(["*"] * i)
    print(spaces + stars)
# Bottom half
for i in range(2, n + 1):
    spaces = " " * (n - i)
    stars = " ".join(["*"] * i)
    print(spaces + stars)
```
