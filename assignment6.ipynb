### 1. Print N Times

```python
n = int(input())
for i in range(n):
    print(0)
```

### 2. Print N Integers

```python
n = int(input())
for i in range(1, n + 1):
    print(i)
```

### 3. First N Natural Numbers

```python
n = int(input())
for i in range(1, n + 1):
    print(i)
```

### 4. Print Integers - 2 (M to N)

```python
m = int(input())
n = int(input())
for i in range(m, n + 1):
    print(i)
```

### 5. Average of N Numbers

```python
n = int(input())
sum_val = 0
for i in range(1, n + 1):
    sum_val += i
average = sum_val / n
print(average)
```

### 6. Cubes of N Numbers

```python
n = int(input())
for i in range(1, n + 1):
    print(i * i * i)
```

### 7. Solid Rectangle

```python
rows = int(input())
cols = int(input())
for i in range(rows):
    row_stars = "* " * cols
    print(row_stars.strip())
```

### 8. 10 Numbers after N

### 9. N Numbers after M

```python
m = int(input())
n = int(input())
for i in range(m + 1, m + n + 1):
    print(i)
```

### 10. Sum of N Numbers from M

```python
m = int(input())
n = int(input())
sum_val = 0
for i in range(m, m + n):
    sum_val += i
print(sum_val)
```

### 11. Sum of N Natural Numbers

### 12. Read N Inputs

```python
n = int(input())
for i in range(n):
    val = int(input())
    print(val)
```

### 13. Product of N Inputs

```python
n = int(input())
product = 1
for i in range(n):
    val = int(input())
    product *= val
print(product)
```

### 14. Sum of Given Numbers
