### 1. Print Length Times

```python
string = input()
length = len(string)
first_character = string[0]

for i in range(length):
    print(first_character)
```

### 3. Square - 2

```python
n = int(input())
for i in range(1, n + 1):
    row = (str(i) + " ") * n
    print(row.strip())
```

### 13. Average of N Inputs

```python
n = int(input())
total_sum = 0
for i in range(n):
    number = int(input())
    total_sum = total_sum + number

average = total_sum / n
print(average)
```

### 14. Product of N numbers after X

```python
x = int(input())
n = int(input())
product = 1

for i in range(1, n + 1):
    product = product * (x + i)

print(product)
```

### 15. Sum of N Squares

```python
n = int(input())
total_sum = 0
for i in range(1, n + 1):
    total_sum = total_sum + (i * i)

print(total_sum)
```
